```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cleaned Healthcare dataset
df = pd.read_excel('cleaned_healthcare_data.xlsx')

# Chart Style
sns.set_theme(style="whitegrid")

# ---------------------------------------------------------
# 1. Age Distribution
# ---------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'], kde=True, color='skyblue', bins=10)
plt.title('1. Age Distribution of Patients')
plt.xlabel('Age')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# 2. Average BMI & Glucose by Disease Stage
# ---------------------------------------------------------
plt.figure(figsize=(8, 5))
disease_metrics = df.groupby('Disease')[['BMI', 'Glucose']].mean()
disease_metrics.plot(kind='bar', figsize=(8, 5), colormap='viridis')
plt.title('2. Average BMI & Glucose Level by Disease Stage')
plt.xlabel('Disease Category')
plt.ylabel('Average Value')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# 3. BP vs Cholesterol (by Smoking Habit)
# ---------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='BP', y='Cholesterol', hue='Smoking', style='Gender', s=100, palette='Set1')
plt.title('3. Blood Pressure vs Cholesterol (by Smoking Status)')
plt.xlabel('Blood Pressure (BP)')
plt.ylabel('Cholesterol')
plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# 4. Disease Count by Gender
# ---------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Disease', hue='Gender', palette='Set2')
plt.title('4. Disease Distribution across Gender')
plt.xlabel('Disease Category')
plt.ylabel('Patient Count')
plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# 5. Correlation Heatmap
# ---------------------------------------------------------
plt.figure(figsize=(8, 6))
corr_matrix = df.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('5. Correlation Heatmap of Healthcare Features')
plt.tight_layout()
plt.show()
```
